```python
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df = pd.read_csv("RentRaw.csv", low_memory=False)
df.info()
df.head(20)
```

# Handling Missing / null values

```python
#To check before handling
df.isnull().sum()
```

```python
# drop missing values records`
df = df.dropna()
df = df.reset_index(drop=True)

df.isnull().sum()
```

```python
# drop BuiltUpSize = 0
df = df[df["BuiltUpSize"] != 0].reset_index(drop=True)

# drop NoOfBedroom = 0
df = df[df["NoOfBedroom"] != 0].reset_index(drop=True)

# drop NoOfBathroom = 0
df = df[df["NoOfBathroom"] != 0].reset_index(drop=True)

# drop NoOfParking = 0
df = df[df["NoOfParking"] != 0].reset_index(drop=True)

# drop RentalPerMth = 0
df = df[df["RentalPerMth"] != 0].reset_index(drop=True)
df.describe()
```

# Handling Duplicates

```python
# Check duplicates ignoring ID
print("{} duplicates".format(df.drop(columns=['ID']).duplicated().sum()))

# Drop records
df = df[~df.drop(columns=['ID']).duplicated()]
df = df.reset_index(drop=True)

print("{} duplicates".format(df.drop(columns=['ID']).duplicated().sum()))
```

# Analysis

Use only selected Features : 

```python
df = df[["BuiltUpSize", "City", "Furnishing", "NoOfBathroom", "NoOfBedroom", "NoOfParking", "State", "PropertyType", "RentalPerMth"]]
df.describe()
```

```python
print("=== NUMERICAL STATISTICS ===")
print(df[["BuiltUpSize", "NoOfBathroom", "NoOfBedroom", "NoOfParking", "RentalPerMth"]].describe().to_string())
print("\n")

# Value counts for categorical data , for readability
print("=== CATEGORICAL STATISTICS ===")
for col in ["Furnishing", "PropertyType", "State", "City"]:
    print(f"--- {col} (Top 10) ---")
    print(df[col].value_counts().head(10).to_string())
    print("\n")

# 3. Create visualizations
fig, axes = plt.subplots(3, 3, figsize=(18, 15))
fig.suptitle('Univariate Analysis of Rental Property Features', fontsize=20, y=1.02)
sns.set_theme(style="whitegrid")

# Plot 1: RentalPerMth
rental_cap = df['RentalPerMth']
sns.histplot(df[df['RentalPerMth'] <= rental_cap]['RentalPerMth'], bins=40, ax=axes[0, 0], kde=True, color='blue')
axes[0, 0].set_title('Rental Per Month')

# Plot 2: BuiltUpSize
size_cap = df['BuiltUpSize']
sns.histplot(df[df['BuiltUpSize'] <= size_cap]['BuiltUpSize'], bins=40, ax=axes[0, 1], kde=True, color='green')
axes[0, 1].set_title('Built Up Size')

# Plot 3: NoOfBedroom
sns.countplot(x=df['NoOfBedroom'], hue=df['NoOfBedroom'], ax=axes[0, 2], palette='viridis', legend=False)
axes[0, 2].set_title('Number of Bedrooms')
axes[0, 2].tick_params(axis='x', rotation=45)

# Plot 4: NoOfBathroom
sns.countplot(x=df['NoOfBathroom'], hue=df['NoOfBathroom'], ax=axes[1, 0], palette='viridis', legend=False)
axes[1, 0].set_title('Number of Bathrooms')

# Plot 5: NoOfParking
sns.countplot(x=df['NoOfParking'], hue=df['NoOfParking'], ax=axes[1, 1], palette='viridis', legend=False)
axes[1, 1].set_title('Number of Parking Spots')

# Plot 6: Furnishing
sns.countplot(y=df['Furnishing'], hue=df['Furnishing'], ax=axes[1, 2], order=df['Furnishing'].value_counts().index, palette='magma', legend=False)
axes[1, 2].set_title('Furnishing Status')

# Plot 7: PropertyType
sns.countplot(y=df['PropertyType'], hue=df['PropertyType'], ax=axes[2, 0], order=df['PropertyType'].value_counts().head(10).index, palette='magma', legend=False)
axes[2, 0].set_title('Property Type (Top 10)')

# Plot 8: State
sns.countplot(y=df['State'], hue=df['State'], ax=axes[2, 1], order=df['State'].value_counts().head(10).index, palette='magma', legend=False)
axes[2, 1].set_title('State Distribution')

# Plot 9: City
sns.countplot(y=df['City'], hue=df['City'], ax=axes[2, 2], order=df['City'].value_counts().head(10).index, palette='magma', legend=False)
axes[2, 2].set_title('City Distribution (Top 10)')

plt.tight_layout()
```

# Correlation

```python
# calculate correlation matrix
corr = df[["BuiltUpSize", "NoOfBathroom", "NoOfBedroom", "NoOfParking", "RentalPerMth"]].corr()
corr

# visualize correlation matrix
plt.subplots(figsize=(8,6))

# Added annot=True to show labels, and fmt=".2f" to round to 2 decimal places
fig = sns.heatmap(corr, 
            xticklabels=corr.columns.values,
            yticklabels=corr.columns.values,
            annot=True,
            fmt=".2f",
            cmap="coolwarm") # Adding a colormap like coolwarm helps distinguish positive/negative

fig.get_figure().savefig("corr_heatmap.png", bbox_inches="tight")
```
