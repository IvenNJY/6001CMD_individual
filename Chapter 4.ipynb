# Import Data

```python
import numpy as np
import pandas as pd
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns

TARGET = "RentalPerMth"
RANDOM_STATE = 42
```

```python
df = pd.read_csv("RentRaw.csv", low_memory=False)
df.info()
df.head(20)
```

# Data Cleaning

## Missing Value Imputation

```python
df.isnull().sum

# drop missing values records`
df = df.dropna()
df = df.reset_index(drop=True)

df.isnull().sum()
```

# Noise Reduction

```python
# drop BuiltUpSize = 0
df = df[df["BuiltUpSize"] != 0].reset_index(drop=True)

# drop NoOfBedroom = 0
df = df[df["NoOfBedroom"] != 0].reset_index(drop=True)

# drop NoOfBathroom = 0
df = df[df["NoOfBathroom"] != 0].reset_index(drop=True)

# drop NoOfParking = 0
df = df[df["NoOfParking"] != 0].reset_index(drop=True)

# drop RentalPerMth = 0
df = df[df["RentalPerMth"] != 0 ].reset_index(drop=True)
df.describe()
```

## Duplicate Removal

```python
duplicates_before = df.drop(columns=['ID']).duplicated().sum()

df = df[~df.drop(columns=['ID']).duplicated()]
df = df.reset_index(drop=True)

duplicates_after = df.drop(columns=['ID']).duplicated().sum()
duplicate_count = duplicates_before - duplicates_after

print(f"Duplicate listings removed: {duplicate_count}")
print(f"Clean shape: {df.shape}")
print(df.isna().sum())
```

# Data Transformation

## Log Transformation

```python
numeric_columns = [
    "BuiltUpSize", "NoOfBedroom", "NoOfBathroom", "NoOfParking", "RentalPerMth"
]

transformation_input = df[numeric_columns].copy()
transformation_input = transformation_input.fillna(transformation_input.median())
built_up_size_before = transformation_input["BuiltUpSize"].copy()

df[numeric_columns] = np.log1p(transformation_input)
built_up_size_after = df["RentalPerMth"]

print("Log transformation applied directly to df:")
display(df.head())

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.histplot(built_up_size_before, bins=30, kde=True, ax=axes[0], color="steelblue")
axes[0].set_title("RentalPerMth Before Log Transformation")
axes[0].set_xlabel("RentalPerMth")
axes[0].set_ylabel("Count")

sns.histplot(built_up_size_after, bins=30, kde=True, ax=axes[1], color="darkorange")
axes[1].set_title("RentalPerMth After Log Transformation")
axes[1].set_xlabel("RentalPerMth (log)")
axes[1].set_ylabel("Count")

plt.tight_layout()
plt.show()
```

# Feature Engineering
## Feature Selection

```python
selected_features = [
    "BuiltUpSize", "City", "Furnishing", "NoOfBathroom", "NoOfBedroom", "NoOfParking", "State", "PropertyType", "RentalPerMth"
]

selected_data = df[selected_features].copy()

print("Selected features:")
print(selected_features)
print("Selected features after preprocessing:")
display(selected_data.head())
print(f"Output shape: {selected_data.shape}")
```

## Encoding Categorical Variables

```python
categorical_columns = ["City", "State", "PropertyType"]

encoded_data = pd.get_dummies(
    selected_data,
    columns=categorical_columns,
    dtype=int
)

print("Categorical variables encoded:")
display(encoded_data.head())
print(f"Encoded output shape: {encoded_data.shape}")
```
